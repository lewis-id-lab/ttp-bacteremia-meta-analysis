**Background.** Blood culture time to positivity (TTP) has been proposed as a prognostic marker in bacteremia, but studies use different thresholds, pathogens, and outcomes. We synthesized the association between short versus long TTP and clinical outcomes.

**Methods.** We searched PubMed, Embase, Scopus, Web of Science, and the Cochrane Library (May 2026), supplemented by an AI-assisted Elicit search, and extracted effect sizes from 100 studies meeting strict eligibility criteria. Log odds ratios for all-cause mortality (short versus long TTP) were pooled with a Bayesian random-effects model. We examined pathogen class, TTP cutpoint, publication bias, and a sensitivity analysis adding studies meeting more inclusive criteria. Persistent bacteremia and microbiological clearance were analyzed as secondary outcomes.

**Results.** Fifty-eight studies (72 effect sizes) reported mortality. Short TTP was associated with higher mortality (pooled OR 2.05; 95% credible interval \[CrI\], 1.66 to 2.57), with substantial heterogeneity (median I² 62%; 95% prediction interval, 0.51 to 8.50). The association was larger in Gram-negative (OR 2.67; 95% CrI, 1.75 to 4.11) than Gram-positive bacteremia (OR 1.47; 95% CrI, 0.83 to 2.50), although the credible intervals overlapped. Higher TTP cutpoints were associated with weaker associations overall, but this was not consistent across subgroups. The funnel plot was asymmetric, and trim-and-fill gave an adjusted OR of 1.51 (95% CI, 1.23 to 1.86). Short TTP was also associated with persistent bacteremia (OR 1.72; 95% CrI, 1.06 to 2.80); data on clearance were sparse.

**Conclusions.** Short TTP is associated with higher mortality, particularly in Gram-negative bacteremia, but heterogeneity, cutpoint dependence, probable confounding, and small-study effects limit its use as a stand-alone prognostic tool. Standardized thresholds and adjusted, individual-patient analyses are needed.

# Introduction

Bloodstream infection is a major cause of morbidity and mortality, and early risk stratification remains a practical clinical problem. Automated continuous-monitoring blood culture systems record the time to positivity (TTP), the interval from the start of incubation to the instrument’s detection of microbial growth. Because growth kinetics depend on the starting inoculum and on organism-specific factors, shorter TTP has been proposed as a readily available, laboratory-generated marker of high bacterial burden and severe infection.

Individual cohorts have examined whether TTP carries prognostic information in specific settings, including *Staphylococcus aureus* bacteremia, where TTP has been explored as a marker of endocarditis and poor outcome ([1](#ref-Sowden2008)–[3](#ref-Stromdahl2025)). Other cohort studies have explored the relationship between TTP and Gram-negative bacteremia, including *Pseudomonas aeruginosa* ([4](#ref-Willmann2013), [5](#ref-Tang2017)) and *Escherichia coli* bloodstream infections ([6](#ref-Martinez2006)), candidemia ([7](#ref-Keighley2023)) as well as its potential as a stewardhwip tool in neutropenic patients ([8](#ref-PuertaAlcalde2019)). These studies differ in organism mix, patient population, outcome definition, and the TTP threshold used to define “short” versus “long” TTP. Whether the association is consistent across pathogens or varies with the chosen cutpoint remain unclear.

We therefore updated a systematic review of the association between blood culture TTP and clinical outcomes in bacteremia. Our primary objective was to estimate the association between short versus long TTP and all-cause mortality using a Bayesian random-effects meta-analysis. Secondary objectives were to examine heterogeneity by pathogen class and by TTP cutpoint, to assess publication bias and the robustness of the findings to study inclusion criteria, and to summarize associations with persistent bacteremia and other markers of clinical failure.

# Materials and Methods

## Protocol and registration

The review is reported according to the PRISMA 2020 statement and the PRISMA-S extension for searches. A protocol was prepared for registration in PROSPERO (CRD420261398553). Deviations from the protocol are listed in the supplementary notebook ([Methods and PRISMA](notebooks/methods.qmd)).

## Search strategy and study selection

Five sources were searched (PubMed, Embase, Scopus, Web of Science, and Cochrane CENTRAL) on May 22, 2026, from inception, with no date or language restriction at the search level. The strategies combined a time-to-positivity block with clinical-outcome blocks, whereas the protocol specified no outcome block. The 57 baseline studies came from an earlier review (a university thesis) whose Elicit-assisted search found 55 of them; the other two (Alvarez 2012 and Peri 2026) were identified by hand-searching. The 2026 database searches identified 6,014 records; after 1,409 duplicates were removed in Covidence, 4,605 unique records remained. CINAHL, trial registers, and preprint servers were not searched, although the protocol specified them, and no reference-list searching or citation tracking was performed for the update. Title and abstract screening was not performed in Covidence. Instead, database records not already in the earlier review were assigned to relevance tiers by keyword, and PubMed abstracts were retrieved for records whose exported abstract was missing, to identify those with a TTP signal; this yielded 44 priority candidates. In parallel, a new search with the AI-assisted tool Elicit returned 445 rows (348 unique DOIs) \[TO COMPLETE: Elicit search date and query\], of which 323 candidates remained after studies already extracted were removed. Together these gave 355 candidate reports (323 from the Elicit search and 32 from the database tiering only), all of which were sought for full-text review. Of these, 53 were not retrieved after pre-screening as ineligible (mainly conference abstracts, non-English papers, and methods-only papers), and 302 full-text reports were assessed (5 additional PDFs duplicated baseline studies). Under the strict criterion 43 reports were included, which added to the 57 baseline studies gave 100 studies; 259 were excluded (104 ineligible design, 65 no TTP measured, 54 no quantitative TTP-outcome association, 29 no clinical outcome, 5 duplicates, 2 not bloodstream infection). A second, inclusive re-screen of the 54 reports excluded only for lacking a quantitative association identified 40 further studies (sensitivity pool, 140 studies in total). Eligibility criteria and the full search strings are in the supplementary appendix ([HTML](appendix-preview.html); [Word](downloads/appendix.docx)) and the supplementary notebook ([Methods and PRISMA](notebooks/methods.qmd)). <a href="#fig-prisma" class="quarto-xref">Figure 1</a> shows study flow.

## Data extraction and risk of bias

Data were extracted from original PDFs into a structured relational workbook with linked tables for study characteristics, population, TTP reporting (continuous, dichotomized, or categorical; units; cutpoints), and outcomes (events and totals by TTP group, reported effect estimates, adjustment status, and outcome definitions) using a purpose built script for Claude Code Opus 5 (Anthropic, Inc). The accuracy of extracted data was verified by an investigators. TTP was recorded in hours. Where a study reported both 2×2 data and an effect estimate, the 2×2 data were preferred. The baseline 57 studies were extracted using claFor the 2026 update, full-text screening and data extraction were performed by a large language model agent working from the full-text PDFs and the written eligibility criteria and extraction codebook. Each of the 43 strict-pool extractions was then independently verified field by field against the source PDF by a separate agent that had not performed the original extraction; this led to 7 corrections before analysis. Studies using TTP only as part of a catheter-related bloodstream infection case definition were excluded from effect-size pooling. The human contribution to screening and extraction decisions (spot-checking, adjudication, or second human review) is described in the AI-use statement below \[confirm\], and we did not perform dual independent human screening or report inter-rater agreement.

Risk of bias was assessed for every included study with the Quality in Prognosis Studies (QUIPS) tool across six domains (participation, attrition, prognostic factor measurement, outcome measurement, confounding, and statistical analysis and reporting), each rated as low, moderate, or high, with an overall rating per study ([notebook](notebooks/rob.qmd)). For the update these ratings were generated by the extraction agent and were not independently re-rated by a second human.

## Statistical analysis

Log odds ratios for mortality (short versus long TTP) were computed with `metafor::escalc()` from 2×2 tables where available, otherwise back-calculated from reported OR, HR, or RR with 95% confidence intervals. Effect sizes were pooled with a Bayesian random-effects model (`brms`; study-level random intercepts) with weakly informative priors ($\mu \sim \mathcal{N}(0,1)$, $\tau \sim \text{Half-Cauchy}(0, 0.5)$). Four chains of 4,000 iterations (1,000 warm-up) were run with `adapt_delta = 0.99`. Convergence was assessed with $\hat{R}$ and effective sample sizes. Secondary analyses examined pathogen class, TTP cutpoint (meta-regression), publication bias, and a sensitivity pool that added studies meeting a more inclusive criterion. All code and data are available in the project repository.

# Results

## Study selection and characteristics

The primary (strict) pool comprised 100 studies, of which 58 reported mortality and contributed 72 mortality effect sizes. Overall risk of bias (QUIPS) was low in 24, moderate in 62, high in 13, and critical in 1 studies ([notebook](notebooks/rob.qmd)).

## Primary analysis

Short TTP was associated with higher mortality (pooled OR 2.05 (95% CrI, 1.66 to 2.57); <a href="#fig-forest" class="quarto-xref">Figure 2</a>, <a href="#tbl-overall" class="quarto-xref">Table 1</a>). The posterior probability that short TTP increases mortality was 100.0%, and heterogeneity was substantial (median $I^2$ = 62%).

## TTP cutpoint

A TTP cutpoint was reported for 42 of 72 mortality effect sizes (29 studies); the remainder had no cutpoint recorded in the extraction database. Cutpoints ranged from 2 to 72 h (median 12 h). Across all effect sizes with a cutpoint, the log odds ratio changed by -0.035 (95% CrI, -0.055 to -0.015) per additional hour, so higher thresholds were associated with weaker associations (<a href="#fig-cutpoint" class="quarto-xref">Figure 4</a>). When the analysis was restricted to cutpoints of 24 h or less, the slope was attenuated and its interval included zero (-0.024 (95% CrI, -0.073 to 0.027)), so the gradient depends on the few studies using long thresholds.

Within pathogen class the pattern differed. Among Gram-negative effect sizes (median cutpoint 10 h) the slope was positive (0.062 (95% CrI, -0.007 to 0.128); <a href="#tbl-gneg-cutpoint" class="quarto-xref">Table 3</a>), whereas among Gram-positive effect sizes it was negative with an interval that included zero (-0.028 (95% CrI, -0.059 to 0.005); <a href="#tbl-gpos-cutpoint" class="quarto-xref">Table 4</a>). The overall negative gradient therefore does not hold within Gram-negative bacteremia, and these class-specific analyses rest on small numbers of effect sizes.

## Publication bias

The funnel plot of mortality effect sizes was asymmetric (<a href="#fig-funnel" class="quarto-xref">Figure 5</a>). Egger’s regression test indicated small-study effects (t = 6.19, df = 70, p \< 0.0001; <a href="#tbl-egger" class="quarto-xref">Table 5</a>), and the regression-based estimate for a study with zero standard error was close to no association (OR 0.99; 95% CI 0.97 to 1.01). Begg’s rank correlation test did not detect asymmetry (p = 0.47). The trim-and-fill method imputed 19 missing studies and gave an adjusted pooled OR of 1.51 (95% CI, 1.23 to 1.86), lower than the unadjusted frequentist estimate (OR 2.05). These tests treat all effect sizes as independent, assume that asymmetry is due to selective reporting, and have limited reliability when heterogeneity is high, so they cannot distinguish publication bias from genuine heterogeneity, such as larger associations in smaller studies that used more extreme or optimized cutpoints. The evidence nevertheless suggests that the pooled association may be overestimated.

## Sensitivity analyses

Adding the studies that met only the inclusive criterion increased the number of studies reporting mortality to 67 (84 effect sizes) and gave a pooled OR of 1.92 (95% CrI, 1.59 to 2.35), similar to the primary estimate (<a href="#fig-sens-inc" class="quarto-xref">Figure 6</a>). Pooled estimates by data source were consistent in direction. Relative to 2×2-derived effect sizes (49 effect sizes), the log odds ratio for effect sizes back-calculated from adjusted reported estimates (15 effect sizes) was -0.26 (95% CrI, -0.69 to 0.17) and for unadjusted reported estimates (8 effect sizes) was -0.11 (95% CrI, -0.60 to 0.39). Both intervals included zero. Directly comparing unadjusted with adjusted reported estimates, the difference in log odds ratio was 0.13 (95% CrI, -0.38 to 0.62). Association strength tended to decrease with publication year (change in log odds ratio per year, -0.04 (95% CrI, -0.08 to 0.00)), with an interval that only just included zero. None of these moderator analyses showed a clear difference, but all were underpowered relative to the observed heterogeneity.

## Clinical failure outcomes

Outcomes other than mortality were reported far less often (<a href="#tbl-counts" class="quarto-xref">Table 6</a>). Persistent bacteremia was the most frequently reported, with 12 effect sizes from 12 studies, followed by microbiological clearance (4 effect sizes, 3 studies). No usable effect sizes were available for relapse (0).

Short TTP was associated with higher odds of persistent bacteremia (pooled OR 1.72 (95% CrI, 1.06 to 2.80); <a href="#fig-pb" class="quarto-xref">Figure 7</a>), with substantial heterogeneity (median $I^2$ 69%). Outcome definitions were not uniform: on review of the outcome definitions, 6 effect sizes defined persistent bacteremia by duration of positive cultures, 3 defined it by endocarditis or valve involvement, and 3 used other definitions. In a meta-regression by definition group, endocarditis-based definitions had a log odds ratio 0.60 higher than duration-based definitions (95% CrI -0.20 to 1.26), which is compatible with no difference given the small number of effect sizes per group. The pooled persistent-bacteremia estimate should therefore be interpreted as an association with a heterogeneous composite of outcomes.

For microbiological clearance, the pooled estimate was too imprecise to interpret (OR 1.63 (95% CrI, 0.24 to 9.55)), reflecting only 4 effect sizes with very large between-study heterogeneity (median $I^2$ 90%).

# Discussion

In this Bayesian random-effects meta-analysis, short TTP was associated with higher mortality in bacteremia (pooled OR 2.05 (95% CrI, 1.66 to 2.57); 58 studies, 72 effect sizes). The posterior probability that the association is in the direction of harm was 100.0%. Between-study heterogeneity was moderate to substantial (median $I^2$ 62%), and the 95% prediction interval for a new study (0.51 to 8.50) is wide and includes no association. The pooled estimate is therefore best read as an average association, not as the effect expected in any particular setting. Adding studies that met only a more inclusive criterion gave a similar estimate (OR 1.92 (95% CrI, 1.59 to 2.35)).

**Pathogen class.** The association was larger in Gram-negative bacteremia (OR 2.67 (95% CrI, 1.75 to 4.11)) than in Gram-positive bacteremia (OR 1.47 (95% CrI, 0.83 to 2.50)), for which the credible interval included 1. The two intervals overlap, so this contrast should be considered hypothesis-generating rather than confirmed. A plausible explanation is that TTP in Gram-positive infection, notably *S. aureus*, is influenced by factors other than burden of disease, such as the source of infection and the presence of endovascular foci, whereas in Gram-negative infection it may track inoculum and host response more directly. This explanation is speculative and was not tested directly here.

**Study design and adjustment.** Effect sizes derived from 2×2 tables (OR 2.28 (95% CrI, 1.74 to 3.05)) were larger than those back-calculated from reported effect estimates (OR 1.71 (95% CrI, 1.25 to 2.40)). Meta-regression by data source did not show a clear difference, and the intervals included zero, so the data are compatible with either similar or lower adjusted associations. Crude 2×2 comparisons may carry more confounding by severity of illness, source, or pathogen, in which case the pooled estimate could overstate the independent prognostic contribution of TTP, but the groups also differ in other respects.

**TTP cutpoint.** Studies defined “short” TTP using different thresholds (median cutpoint 12 h), and 30 of 72 mortality effect sizes did not have a reported cutpoint. In meta-regression among the remaining effect sizes, the log OR decreased by 0.035 per additional hour of cutpoint (95% CrI 0.015 to 0.055), meaning that stricter (shorter) thresholds were associated with larger odds ratios overall (<a href="#fig-cutpoint" class="quarto-xref">Figure 4</a>). This gradient was not robust: it was attenuated when cutpoints above 24 h were excluded, and it was reversed within Gram-negative studies. Because thresholds are often chosen post hoc to maximize discrimination in the study sample, the gradient may partly reflect optimistic cutpoint selection or differences in pathogen mix, and it should not be interpreted as validating any single clinical threshold.

**Other outcomes.** For persistent bacteremia, short TTP was associated with higher odds (OR 1.72 (95% CrI, 1.06 to 2.80)), but outcome definitions varied widely between studies. Evidence for microbiological clearance was too sparse and imprecise to interpret (OR 1.63 (95% CrI, 0.24 to 9.55)) ([notebook](notebooks/clinical_failure.qmd)).

**Strengths and limitations.** Strengths include a multi-database search with a May 2026 update, a distinction between strict and inclusive study pools, a Bayesian framework that yields prediction intervals and direct probability statements, and a fully reproducible code base. Limitations are substantial. Most included studies were retrospective cohorts and risk of bias was rated moderate or worse in most (76 of 100 studies). Several studies contributed multiple effect sizes, which we handled with study-level random intercepts but which do not remove all dependence. The TTP threshold, outcome timing, and adjustment sets were not standardized, and adjustment status varied across estimates. Blood culture system, bottle type, transport delay, and prior antibiotic exposure affect TTP and were incompletely reported, so residual confounding by measurement conditions is likely. The funnel plot was asymmetric and Egger’s test indicated small-study effects (<a href="#fig-funnel" class="quarto-xref">Figure 5</a>, <a href="#tbl-egger" class="quarto-xref">Table 5</a>). Trim-and-fill reduced the pooled estimate to OR 1.51, so the primary estimate may overstate the association, although asymmetry can also arise from heterogeneity and cutpoint optimization, and these tests do not account for multiple effect sizes per study. Finally, the meta-analysis is based on aggregate data, which prevents examination of TTP as a continuous predictor within patients.

**Implications.** Short TTP is associated with higher mortality on average, but heterogeneity, the dependence on cutpoint, and probable confounding limit its use as a stand-alone prognostic tool. These results support using TTP as one contextual signal, particularly in Gram-negative bacteremia, alongside clinical severity and pathogen identification. Individual-patient-data analyses with standardized TTP measurement, and prospective validation of pathogen-specific thresholds, are needed before TTP can be used to guide treatment decisions.
